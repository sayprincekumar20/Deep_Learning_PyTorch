4. Matrix Operation

```python
f= torch.randint(size=(2,3), low=0, high=10)
g= torch.randint(size=(3,2),low=0,high=10)
print("f:",f)
print("g:",g)
```

```python
# Matrix multiplication
torch.matmul(f,g)
```

```python
ff= torch.randint(size=(2,3), low=0, high=10)
gg= torch.randint(size=(2,3),low=0,high=10)
print("ff:",ff)
print("gg:",gg)
 # matrix multiplication with Transpose of ff

torch.matmul(ff.T,gg)
```

```python
vector1=torch.tensor([1,2,3])
vector2=torch.tensor([4,5,6])
```

```python
# Dot Product
torch.dot(vector1,vector2)
```
